```python
from pathlib import Path
import csv

student_id = "TUPM-25-0384"

folder = Path.home() / "Documents" / "Activity_5_Files"

txt_files = list(folder.glob("*.txt"))

summary = []

for file in txt_files:

    text = file.read_text()

    lines = [line for line in text.splitlines() if line.strip()]

    words = text.split()
    characters = len(text)

    total_lines = len(lines)
    total_words = len(words)

    words_per_line = total_words / total_lines if total_lines else 0
    chars_per_word = characters / total_words if total_words else 0

    summary.append([
        file.name,
        total_lines,
        total_words,
        characters,
        round(words_per_line, 2),
        round(chars_per_word, 2)
    ])

summary.sort(key=lambda x: x[4], reverse=True)

print("Filename | Lines | Words | Characters | Words/Line | Chars/Word")
print("-" * 75)

for row in summary:
    print(row)

print("\nMost Content-Dense File:", summary[0][0])
print("Least Content-Dense File:", summary[-1][0])

csv_file = folder / "content_summary.csv"

with csv_file.open("w", newline="") as f:

    writer = csv.writer(f)

    writer.writerow([
        "Filename",
        "Lines",
        "Words",
        "Characters",
        "Words/Line",
        "Chars/Word"
    ])

    writer.writerows(summary)

print("\nSummary exported to:", csv_file)
```

```python
from pathlib import Path
from collections import Counter
import string
import csv

student_id = "TUPM-25-0384"

folder = Path.home() / "Documents" / "Activity_5_Files"

txt_files = list(folder.glob("*.txt"))

stopwords = {
    "the", "is", "and", "to",
    "a", "of", "in", "for", "this"
}

word_counter = Counter()

for file in txt_files:

    text = file.read_text().lower()

    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    words = text.split()

    filtered_words = [
        word for word in words
        if word not in stopwords
    ]

    word_counter.update(filtered_words)

print("Most Frequent Meaningful Words:")
print("-" * 40)

for word, count in word_counter.most_common(10):
    print(word, ":", count)

csv_file = folder / f"word_frequency_{student_id}.csv"

with csv_file.open("w", newline="") as f:

    writer = csv.writer(f)

    writer.writerow(["Word", "Count"])

    for word, count in word_counter.items():
        writer.writerow([word, count])

print("\nWord frequency exported to:", csv_file)
```

```python
from pathlib import Path
from datetime import datetime
import shutil

student_id = "TUPM-25-0384"
student_name = "Brylle Edsil F. Untalasco"

base_dir = Path.home() / "Documents" / "Activity_5_Exercises"

base_dir.mkdir(parents=True, exist_ok=True)

source_file = base_dir / "intro_file.txt"

if not source_file.exists():
    source_file.write_text(
        "Sample backup content for Activity 5."
    )

backup_dir = base_dir / f"backup_{student_id}"

backup_dir.mkdir(parents=True, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

backup_filename = (
    f"{source_file.stem}_{student_id}_{timestamp}.txt"
)

backup_path = backup_dir / backup_filename

shutil.copy(source_file, backup_path)

log_file = backup_dir / f"backup_log_{student_id}.txt"

file_size = backup_path.stat().st_size

with log_file.open("a") as log:

    log.write(f"Student ID: {student_id}\n")
    log.write(f"Student Name: {student_name}\n")
    log.write(f"Original File: {source_file.name}\n")
    log.write(f"Timestamp: {timestamp}\n")
    log.write(f"File Size: {file_size} bytes\n")
    log.write(f"Backup Path: {backup_path}\n")
    log.write("-" * 50 + "\n")

print(f"Backup completed for {student_id} ({student_name})")
print(f"File saved as: {backup_filename}")
print(f"Log updated at: {backup_dir}")
```
